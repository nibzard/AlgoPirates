```python
# Pokreni me! Pogledaj današnji papirni list za plan.

# Izazov gradnje tornja - Vježba petlji i množenja
# Ovaj program simulira gradnju tornja korak po korak
# Vježbat ćemo množenje kroz ponavljano zbrajanje

print("🏗️ Gradimo tvoj toranj korak po korak!")
print("Svaki korak dodaje blokove da toranj bude viši")
print()

# Počnemo s praznim tornjem (visina = 0)
visina = 0

# Gradimo toranj u 5 koraka
for korak in range(5):
    # Svaki korak dodaje 2 bloka (možeš to promijeniti!)
    dodani_blokovi = 2
    visina += dodani_blokovi
    
    # Prikaži napredak nakon svakog koraka
    print(f'Korak {korak + 1}: Dodano {dodani_blokovi} blokova, ukupna visina: {visina}')

print()
print(f"🎉 Toranj završen! Konačna visina: {visina} blokova")
print(f"Dodali smo {dodani_blokovi} blokova po koraku × 5 koraka = {dodani_blokovi * 5} ukupno blokova")

### ZADATAK ###
# 1. Promijeni 'dodani_blokovi' da vježbaš različite tablice množenja
# 2. Pokušaj promijeniti range(5) da napraviš više ili manje koraka
# 3. Možeš li predvidjeti konačnu visinu prije pokretanja?
```
